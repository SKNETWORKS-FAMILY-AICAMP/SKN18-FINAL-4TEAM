```python
# Re-initialize df_tran_no from df_tran_no_error after dropping NaNs
df_tran_no = df_tran_no_error.dropna(subset=['transcript'])

# Install langdetect if not already installed (run once per session)
try:
    import langdetect
except ImportError:
    !pip install langdetect
    import langdetect

from langdetect import detect, DetectorFactory
# Ensure consistent results for langdetect
DetectorFactory.seed = 0

def detect_language(text):
    try:
        return detect(text)
    except:
        return 'unknown' # Handle cases where language detection might fail (e.g., very short strings or errors)

# Apply language detection to the 'transcript' column
df_tran_no['detected_lang'] = df_tran_no['transcript'].apply(detect_language)

# Filter out rows where the detected language is not Korean ('ko') or English ('en')
df_tran_no = df_tran_no[df_tran_no['detected_lang'].isin(['ko', 'en'])]

print(f"Number of rows after filtering by detected language: {len(df_tran_no)}")
```

```python
# final_data.csv 에서 특정 row 삭제
df = pd.read_csv("final_data.csv")
#전체 row가 빈 row 삭제
df = df.dropna(how='all')
df.to_csv('final_data.csv', index=False)
print("DataFrame 'df' saved to 'final_data.csv' after removing empty rows")
```

```python
# domain 컬럼이 빈 값이면 algorithm 으로 채우기
df = pd.read_csv("final_data.csv")
df['domain'].fillna('algorithm', inplace=True)
df.to_csv('final_data.csv', index=False)
print("DataFrame 'df' saved to 'final_data.csv' after filling empty 'domain' values with 'algorithm'")
```

```python
# domain 기준으로 domain에 들어간 값이 같은 것끼리 재정렬
df = pd.read_csv("final_data.csv")
df = df.sort_values(by=['domain']).reset_index(drop=True)
# id 값 지우고 다시 순서대로
df = df.drop(columns=['id'])
df.to_csv('final_data.csv', index=False)
print("DataFrame 'df' saved to 'final_data.csv' after sorting by 'domain' and resetting 'id'")
```

```python
# domain 컬럼이 livecoding인 경우 category 기준으로 재정렬, 원본 csv 파일을 변경하는거임
df = pd.read_csv("final_data.csv")
df_livecoding = df[df['domain'] == 'livecoding']
df_livecoding = df_livecoding.sort_values(by=['category']).reset_index(drop=True)
```

```python
# domain이 algorithm인 경우의 row들은 그대로 내버려 두고, domain이 livecoding인 경우의 row들만 category 기준으로 재정렬한 후에 다시 합치기
df_algorithm = df[df['domain'] == 'algorithm']
df_algorithm = df_algorithm.reset_index(drop=True)
df_final = pd.concat([df_algorithm, df_livecoding], ignore_index=True)
# id 다시 붙이기
df_final['id'] = range(1, len(df_final) + 1)
df_final.to_csv('final_data.csv', index=False)
print("DataFrame 'df_final' saved to 'final_data.csv' after merging sorted 'livecoding' and 'algorithm' domains")

# 제대로 된거 맞냐
df_check = pd.read_csv("final_data.csv")
df_check.head()
```

```python
#final_data5.csv에서 category 컬럼 안에 있는 값들이 []로 감싸져 있지 않은 경우 []로 감싸지도록 변경하기
df_final_data5 = pd.read_csv("final_data5.csv")
def ensure_list_format(category_value):
    if isinstance(category_value, str) and not (category_value.startswith('[') and category_value.endswith(']')):
        return f"['{category_value}']"
    return category_value
df_final_data5['category'] = df_final_data5['category'].apply(ensure_list_format)
df_final_data5.to_csv('final_data5.csv', index=False)
print("DataFrame 'df_final_data5' saved to 'final_data5.csv' after updating 'category' column format")  
```

```python
import pandas as pd
```

```python
# 1. CSV 파일 불러오기
df = pd.read_csv('final_data_with_summary.csv')
```

```python
# 2. 삭제할 컬럼 리스트 정의
columns_to_drop = ['duration', 'transcript_source', 'relevance_score', 'transcript']
```

```python
# 3. 컬럼 삭제 (axis=1은 열을 의미)
df_dropped = df.drop(columns=columns_to_drop)
```

```python
# 4. 결과를 새로운 CSV 파일로 저장
df_dropped.to_csv('final_data_cleaned.csv', index=False)
```
